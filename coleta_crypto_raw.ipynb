```python
import psycopg2
import os
from dotenv import load_dotenv

load_dotenv()
user = os.getenv("user")
password = os.getenv("password")
port = os.getenv("port")


try:
    conexao = psycopg2.connect(
        host = "localhost",
        database = "crypto_analytics",
        user = user,
        password = password,
        port = port
    )
    print("Conectado com sucesso!")
except Exception as e:
    print(repr(e))
```
